```python
columns = ['unit', 'duration', 'halfwidth', 'pt_ratio', 'rep_slope', 'rec_slope', 'amp', 'spread', 'vel_above', 'vel_below']

waveform_metrics = {column: [] for column in columns}

for unit, data_df in [(73, data_073_df), (695, data_695_df)]:

    for probe_num in range(num_probes):
        # waveform_metrics['probe_num'].append(probe_num)
        waveform_metrics['unit'].append(unit)

        probe_df = data_073_df[data_073_df.probe_num == probe_num]

        max_channel = int(utils.process.get_probe_max_channel(probe_df))
        include_channels = [i for i in range(max_channel-10, max_channel+11)]
        # waveform_metrics['max_channel'].append(max_channel)

        times = probe_df[probe_df.channel_num == max_channel].t.values[0]
        single_waveform = probe_df[probe_df.channel_num == max_channel].ve.values[0]

        waveform_metrics['duration'].append(float(utils.waveform_metrics.calculate_waveform_duration(single_waveform, times)))
        waveform_metrics['halfwidth'].append(float(utils.waveform_metrics.calculate_waveform_halfwidth(single_waveform, times)))
        waveform_metrics['pt_ratio'].append(float(utils.waveform_metrics.calculate_waveform_PT_ratio(single_waveform)))
        waveform_metrics['rep_slope'].append(float(utils.waveform_metrics.calculate_waveform_repolarization_slope(single_waveform, times)))
        waveform_metrics['rec_slope'].append(float(utils.waveform_metrics.calculate_waveform_recovery_slope(single_waveform, times)))

        multi_waveforms = []
        for include_channel in include_channels:
            multi_waveforms.append(probe_df[probe_df.channel_num == include_channel].ve.values[0])
        multi_waveforms = np.array(multi_waveforms)

        amp, spread, vel_above, vel_below = utils.waveform_metrics.calculate_2D_features(multi_waveforms, times, include_channels.index(max_channel))
        waveform_metrics['amp'].append(float(amp))
        waveform_metrics['spread'].append(float(spread))
        waveform_metrics['vel_above'].append(float(vel_above))
        waveform_metrics['vel_below'].append(float(vel_below))



```

```python
wm_df = pd.DataFrame.from_dict(waveform_metrics)  #, orient='index')
wm_df.head()
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
data = wm_df.loc[:,columns].values
```

```python
duration = data[:,0]
halfwidth = data[:,1]

scatter = plt.scatter(data[:,0], data[:,2], c=wm_df.unit)
plt.legend(*scatter.legend_elements(), loc='upper right')
plt.xlabel('duration')
plt.ylabel('halfwidth')
```

```python
kmeans_model = KMeans(n_clusters=2)

y_pred = kmeans_model.fit_predict(data)
```

```python
scatter = plt.scatter(data[:,0], data[:,2], c=y_pred)
plt.legend(*scatter.legend_elements(), loc='upper right')
```
