```python
import pickle
import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from mpl_toolkits.axes_grid1.anchored_artists import AnchoredSizeBar
from matplotlib_scalebar.scalebar import ScaleBar
```

## SOMA

```python
syns_type = 'soma'
soma_num_syns = [40, 70, 100]

soma_sims_data = {num_syn: {} for num_syn in soma_num_syns}

for i, num_syn in enumerate(soma_num_syns):

    sim_label = f'FR26_{syns_type}-0nA-{num_syn}Ex1-1x40Hz-active_NEW'

    output_dir = os.path.join('/home/kedoxey/PyramidalCellSimulations','models/NEURON/NMLCL000073-NEURON','output')
    sim_fname = f'L5PC_{sim_label}_data.pkl'
    sim_path = os.path.join(output_dir, f'FR26_{syns_type}', sim_label, sim_fname)

    with open(sim_path,'rb') as fp:
        all_data = pickle.load(fp)

    simData = all_data['simData']

    soma_sims_data[num_syn]['t'] = np.array(simData['t'])
    soma_sims_data[num_syn]['V_soma'] = np.array(simData['V_soma_0']['cell_0'])
```

```python

# num_syns = [50, 200, 450]
%matplotlib inline
fig, axs = plt.subplots(3,1,figsize=(8,5.5))
axs = axs.ravel()

colors = ['tab:pink', 'tab:olive', 'grey']

for i, num_syn in enumerate(soma_num_syns):

    t = soma_sims_data[num_syn]['t']
    V_soma = soma_sims_data[num_syn]['V_soma']

    axs[i].plot(t, V_soma, zorder=12, label=num_syn, color=colors[i])
    axs[i].set_yticks([-80, -60, -40, -20, 0, 20])
    # axs[i].set_title(f'{num_syn} synapses')
    axs[i].spines['right'].set_visible(False)
    axs[i].spines['top'].set_visible(False)
    axs[i].spines['bottom'].set_visible(False)
    # axs[i].axvline(300)
    # axs[i].text(1.01,0.5, f"{num_syn}", size=10, ha="center", transform=axs[i].transAxes)


    scalebar = AnchoredSizeBar(axs[i].transData,
                                1, '40 mV', 'lower right', 
                                pad=0.1,
                                color='black',
                                frameon=False,
                                size_vertical=40)

    # axs[i].add_artist(scalebar)


axs[1].set_ylabel('Voltage (mV)')
axs[0].set_xticks([])
axs[1].set_xticks([])
axs[-1].set_xlabel('Time (ms)')
axs[-1].spines['bottom'].set_visible(True)

# axs[0].text(1.01,1.5, 'Number of Synapses', size=12, ha="center", transform=axs[0].transAxes)

# fig.text(1, 0.95, "Number of Synapses", fontsize=10, ha='center')

fig.legend(loc='upper right')

fig.suptitle('Soma Memebrane Potential')
fig.tight_layout()
# fig.savefig(os.path.join(output_dir, 'firing_rates', 'soma_time_series.png'),dpi=300)
```

# Combined

```python
syn_colors = {'soma': '#26aaad', 'basal': '#b12625', 'apical_distal': '#e38925'}
```

```python
output_dir = os.path.join('/home/kedoxey/PyramidalCellSimulations','models/NEURON/NMLCL000073-NEURON','output')
firing_rates_fname = 'firing_rates-2026.pkl'
firing_rates_path = os.path.join(output_dir,'firing_rates', firing_rates_fname)

with open(firing_rates_path,'rb') as fp:
    firing_rates = pickle.load(fp)
```

```python
fig, axs = plt.subplots(1,3,figsize=(15,5), sharey=True)
axs = axs.ravel()

xlims = [400, 400, 4000]

axs[0].axvline(100, color='k', alpha=0.4, label='chosen')
axs[1].axvline(100, color='k', alpha=0.4, label='chosen')
axs[2].axvline(1400, color='k', alpha=0.4, label='chosen')

for i, syns_type in enumerate(['soma', 'basal', 'apical_distal']):
    
    num_syns = np.array(list(firing_rates[syns_type].keys()))
    sorted_syns = np.argsort(num_syns)
    syns_frs = np.array(list(firing_rates[syns_type].values()))

    avg_num_syns = num_syns[np.abs(syns_frs - np.average(syns_frs[syns_frs != 0])).argmin()]
    print(f'{syns_type} average at {avg_num_syns} synapses')

    med_num_syns = num_syns[np.abs(syns_frs - np.median(syns_frs[syns_frs != 0])).argmin()]
    print(f'{syns_type} median at {med_num_syns} synapses')

    axs[i].axvline(avg_num_syns, color='tab:pink', label='average')
    axs[i].axvline(med_num_syns, color='tab:olive', label='median')
    axs[i].plot(num_syns[sorted_syns], syns_frs[sorted_syns], color=syn_colors[syns_type])
    axs[i].set_title(syns_type)
    axs[i].set_xlim([0,xlims[i]])
    axs[i].legend(loc='upper right')
    axs[i].set_xlabel('Number of Synapses')

axs[0].set_ylabel('Evoked Firing Rate (Hz)')
axs[0].set_yticks([0, 4, 8, 12, 16, 20, 24])
axs[0].set_ylim(bottom=0)


fig.tight_layout()
```
